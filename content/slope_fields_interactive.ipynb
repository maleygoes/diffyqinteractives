# Slope Fields: Seeing an ODE Before Solving It

**Time:** about 15–20 minutes  
**Goal:** Build intuition for what a differential equation tells us *locally* and how those local slopes combine into solution curves.

A first-order differential equation

$$
\frac{dy}{dt}=f(t,y)
$$

does **not** directly tell us the value of \(y\). Instead, at every point \((t,y)\), it tells us the slope a solution would have if it passed through that point.

A **slope field** draws many tiny line segments with those slopes.

> **Main idea:** a solution curve is a path that follows the little line segments.


## 1. First look: what does $y'=y$ mean?

Before running anything, make a prediction.

For

$$
y'=y,
$$

answer mentally:

- What should the slopes look like when $y>0$?
- What should happen when $y<0$?
- What happens along $y=0$?

Then run the cell below.


```python
import numpy as np
import matplotlib.pyplot as plt


def slope_field(f, t_range=(-3, 3), y_range=(-3, 3), density=21, solutions=None):
    t = np.linspace(*t_range, density)
    y = np.linspace(*y_range, density)
    T, Y = np.meshgrid(t, y)
    M = f(T, Y)

    # Normalize segment lengths so steep slopes do not dominate the picture.
    U = 1 / np.sqrt(1 + M**2)
    V = M / np.sqrt(1 + M**2)

    plt.figure(figsize=(8, 5))
    plt.quiver(T, Y, U, V, angles='xy', pivot='middle', headlength=0, headwidth=0, headaxislength=0)
    plt.axhline(0, linewidth=0.8)
    plt.axvline(0, linewidth=0.8)
    plt.xlim(t_range)
    plt.ylim(y_range)
    plt.xlabel('t')
    plt.ylabel('y')
    plt.grid(alpha=0.2)

    if solutions:
        for ts, ys, label in solutions:
            plt.plot(ts, ys, linewidth=2, label=label)
        plt.legend()

    plt.show()

f = lambda t, y: y
slope_field(f)
```

### Check your intuition

Look across any horizontal row of the graph.

Because $y'=y$ depends only on $y$, every point with the same $y$-value has the same slope.

Write a one-sentence answer in your notes:

**Why are the slopes horizontal at $y=0$, positive above it, and negative below it?**


## 2. Trace possible solution curves

A slope field contains many possible solutions. The initial condition chooses one of them.

Run the next cell. It overlays several exact solutions of $y'=y$.


```python
t = np.linspace(-3, 3, 400)
solutions = []
for y0 in [-2, -1, 0.5, 1, 2]:
    y = y0 * np.exp(t)
    solutions.append((t, y, f'y(0)={y0}'))

slope_field(lambda t, y: y, solutions=solutions)
```

### Observe

Notice that the curves do not cut across the little segments; they flow along them.

Answer briefly:

1. Which initial condition gives an equilibrium solution?
2. If $y(0)>0$, does the solution move upward or downward as $t$ increases?
3. If $y(0)<0$, what happens?

You do **not** need the formula $y=Ce^t$ to make these predictions from the slope field.


## 3. A parameter you can change: $y'=a-y$

Now consider

$$
y'=a-y.
$$

The parameter $a$ changes where the equilibrium occurs.

Before running the interactive cell, predict:

- For what value of $y$ will the slope be zero?
- Above that value, should solutions rise or fall?
- Below it, should solutions rise or fall?


```python
from ipywidgets import interact, FloatSlider


def explore_equilibrium(a=1.0):
    f = lambda t, y: a - y
    slope_field(f, t_range=(-3, 3), y_range=(-3, 5))
    print(f'Equilibrium level: y = {a:.1f}')

interact(
    explore_equilibrium,
    a=FloatSlider(value=1.0, min=-2.0, max=4.0, step=0.5, description='a')
);
```

### What changes? What stays the same?

Move the slider through several values.

In your notes, complete these statements:

- The equilibrium occurs where \(y=\) ________.
- Above equilibrium, the slopes are __________.
- Below equilibrium, the slopes are __________.
- Therefore solutions are pushed __________ the equilibrium.

This is an example of a **stable equilibrium**.


## 4. When the slope depends on both variables

Now examine

$$
y'=t-y.
$$

This is different. The slope changes when either \(t\) or \(y\) changes.

Before running the plot, find the points where the slope is zero:

$$
t-y=0.
$$

What line in the $(t,y)$-plane is that?


```python
slope_field(lambda t, y: t - y, t_range=(-3, 3), y_range=(-3, 3), density=25)
```

### Find the zero-slope line

Look for where the line segments are horizontal.

You should see them approximately along

$$
y=t.
$$

Now inspect regions on either side of that line.

- Where $t-y>0$, slopes are positive.
- Where $t-y<0$, slopes are negative.

This gives us qualitative information without solving the ODE.


## 5. Predict a solution before seeing it

Suppose

$$
y'=t-y, \qquad y(0)=2.
$$

Before running the next cell, use only the slope field to predict:

1. Is the initial slope positive or negative?
2. Will the solution initially move up or down?
3. Will it keep moving that way forever, or might its behavior change?

Make your prediction first.


```python
from scipy.integrate import solve_ivp


def rhs(t, y):
    return t - y[0]

sol = solve_ivp(rhs, (-3, 3), [2], dense_output=True, max_step=0.05)
ts = np.linspace(-3, 3, 400)
ys = sol.sol(ts)[0]

slope_field(
    lambda t, y: t - y,
    t_range=(-3, 3),
    y_range=(-3, 4),
    density=25,
    solutions=[(ts, ys, 'solution through y(0)=2')]
)
```

### Interpretation

The important point is not the numerical solver. Focus on the geometry.

The solution begins by following the local slopes. As it moves through the plane, the slopes it encounters change.

That is why a solution can change direction even though the differential equation itself has not changed.
