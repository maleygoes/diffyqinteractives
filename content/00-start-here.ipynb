## 3. Cells that draw

Many cells you will run in these activities produces a picture instead of a number. Run the two

```python
%pip install -q ipywidgets
```

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 10, 300)

plt.figure(figsize=(7, 4))
plt.plot(t, np.exp(-0.3 * t) * np.cos(2 * t), linewidth=2)
plt.title('A damped oscillation (you will meet this in Chapter 5)')
plt.xlabel('t')
plt.ylabel('y')
plt.grid(alpha=0.3)
plt.show()
```

## 4. Sliders

Some cells give you something to drag. Run the one below, then move the slider slowly and watch
what happens to the curve.

```python
from ipywidgets import interact, FloatSlider

def show_damping(damping=0.3):
    t = np.linspace(0, 10, 300)
    plt.figure(figsize=(7, 4))
    plt.plot(t, np.exp(-damping * t) * np.cos(2 * t), linewidth=2)
    plt.title(f'damping = {damping:.2f}')
    plt.xlabel('t')
    plt.ylabel('y')
    plt.ylim(-1.1, 1.1)
    plt.grid(alpha=0.3)
    plt.show()

interact(
    show_damping,
    damping=FloatSlider(value=0.3, min=0.0, max=1.0, step=0.05,
                        description='damping', continuous_update=False)
);
```

The picture redraws a moment after you let go of the slider. It is not instant, and that is fine.

When an activity hands you a slider, the useful move is to ask **what changed and what stayed the
same** as you dragged it. Set it to the extremes on purpose. The ends of the range can tell
you more than the middle.

## 5. Order matters

Cells share one memory. A cell can only use something that an *earlier* cell already set up, and
"earlier" means earlier in the order you ran them, not the order they sit on the page.

This is the source of essentially every error you will hit here. Which brings us to the last
thing.

## 6. Errors are survivable

The cell below is broken **on purpose**. Run it. You will get a wall of red text.

```python
print(this_was_never_defined)
```

Nothing is damaged. Nothing is graded. You did not break the site.

Two habits handle almost all of it:

1. **Read the bottom line of the red text, not the top.** The last line is the actual complaint.
   Here it says something like `NameError: name 'this_was_never_defined' is not defined`, which
   means: you asked for something that was never set up.
2. **When in doubt, start over from the top.** Go to the **Kernel** menu and choose
   **Restart Kernel and Run All Cells**. That clears the memory and runs the page in order from
   the beginning. It fixes most stuck notebooks.

If a notebook is truly mangled, close it, right click its name in the file list on the left,
delete it, and refresh the page. A clean copy comes back.
